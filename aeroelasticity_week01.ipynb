```python
# Lab 1A: run each line, then change the numbers and run again.
rho = 1.225        # air density at sea level, kg/m^3
V = 30           # speed, m/s
S = 0.30           # wing area, m^2
CL = 0.6           # lift coefficient (no units)

q = 0.5 * rho * V**2       # dynamic pressure, Pa   (** means "to the power of")
L = q * S * CL             # lift, N
print("dynamic pressure q =", round(q, 2), "Pa")
print("lift L =", round(L, 2), "N")
print("mass it can carry =", round(L / 9.81, 3), "kg")
```

```python
import numpy as np
import matplotlib.pyplot as plt

rho, S, CL = 1.225, 0.30, 0.6
V = np.linspace(0, 30, 61)          # 61 speeds from 0 to 30 m/s
L = 0.5 * rho * V**2 * S * CL       # numpy does the whole array at once

plt.plot(V, L, label="lift")
plt.axhline(2.0 * 9.81, color="r", ls="--", label="weight of a 2 kg plane")
plt.xlabel("speed V (m/s)"); plt.ylabel("lift L (N)")
plt.title("Lift grows with V squared"); plt.grid(True); plt.legend()
plt.show()

# Find the speed where lift equals the weight
W = 2.0 * 9.81
V_needed = np.sqrt(2 * W / (rho * S * CL))
print("speed needed:", round(V_needed, 2), "m/s")
```
